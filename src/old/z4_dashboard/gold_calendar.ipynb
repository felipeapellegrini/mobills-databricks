```python
%sql
use catalog mobills;

with dt_orcamentos as (

    select 
        min(data) as min_dt,
        max(data) as max_dt 
    from gold.orcamentos

),

dt_transacoes as (

    select 
        min(data) as min_dt,
        max(data) as max_dt
    from gold.transacoes

),

union_dt as (

    select 
        min(min_dt) as min_dt,
        max(max_dt) as max_dt
    from (
        select min_dt, max_dt from dt_orcamentos
        union all
        select min_dt, max_dt from dt_transacoes
    )

),

calendar as (

    select 
        EXPLODE(SEQUENCE(min_dt, max_dt, INTERVAL 1 DAY)) AS calendar_date
    from union_dt

)

select 
    calendar_date,
    year(calendar_date) as year,
    month(calendar_date) as month,
    concat(year(calendar_date), lpad(month(calendar_date), 3, '-0'), '-01') as year_month
from calendar
```

```python
(
    _sqldf
    .write
    .mode('overwrite')
    .option('overwriteSchema', 'true')
    .saveAsTable("mobills.gold.calendar")
)
```
