```python
# clear src folders
def clear_source(table_name):
    path = []
    objetos = dbutils.fs.ls(f'/Volumes/mobills/default/raw_data/{table_name}')

    for obj in objetos:
        if obj.path.endswith('.csv'):
            path.append(obj.path)

    for file in path:
        dbutils.fs.rm(file, True)

for t in ['transacoes_pendentes', 'transacoes_confirmadas', 'orcamentos', 'contas', 'cartoes', 'categorias']:
    try:
        clear_source(t)
    except:
        print(f'Pasta {t} vazia')
        continue
```
