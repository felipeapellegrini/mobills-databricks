```python
from pyspark.sql.functions import col
```

```python
spark.sql("""
    GRANT USE CATALOG ON CATALOG mobills_prod TO `data_engineers`
""")
```

```python
schemas = spark.sql("""
SHOW SCHEMAS IN mobills_prod
""")

schemas = schemas.filter(~col('databaseName').isin(['default', 'information_schema'])).collect()

for schema in schemas:
    print(schema['databaseName'])
    spark.sql(f"""
            GRANT USE SCHEMA, SELECT, MODIFY
            ON SCHEMA mobills_prod.{schema['databaseName']}
            TO `data_engineers`
    """)
    tables = spark.sql(f"SHOW TABLES IN mobills_prod.{schema['databaseName']}").collect()
    for table in tables:
        print(f"{table['database']}.{table['tableName']}")
        spark.sql(f"""
                GRANT SELECT
                ON TABLE mobills_prod.{table['database']}.{table['tableName']}
                TO `data_engineers`
        """)
    
```
