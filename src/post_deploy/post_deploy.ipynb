```python
import yaml
```

```python


with open("./grants.yml") as f:
    config = yaml.safe_load(f)

for catalog in config["catalogs"]:
    for grant in catalog["grants"]:
        sql = f"""
        GRANT {",".join(grant["privileges"])}
        ON CATALOG {catalog["name"]}
        TO `{grant["principal"]}`
        """

        spark.sql(sql)

for schema in config["schemas"]:
    for grant in schema["grants"]:
        sql = f"""
        GRANT {",".join(grant["privileges"])}
        ON SCHEMA {schema["name"]}
        TO `{grant["principal"]}`
        """

        spark.sql(sql)
```
